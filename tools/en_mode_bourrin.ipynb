```python
import bs4
import requests as rq
import pandas as pd
```

```python
# Des qu'on doit retrouver un lieu, il faut connaitre le nom de la commune ou l'acte a eu lieu.
# Les communes (anciennes et modernes) sont connues via des ID numériques. Il est donc nécessaire
# de récupérer cet id en faisant un appel au webservice qui les sert:
places_url = "https://agatha.arch.be/search/genealogie/ajax.php"

request = {
    "method": "getMunicipalities",
    "type"  : "CS",         # "CS" si on veut l'état civil apres 1796
    "term"  : "sterrebeek", # location qu'on veut trouver
}

# get
r = rq.post(places_url, data=request)
r.json()
```

```python
# Apparemment, il y a aussi un webservice qui permet de faire la recherche comme ca:
places_url = "https://agatha.arch.be/search/genealogie/ajax.php"

request = {
    "method"     : "getResults",
    "pageSize"   : "10",         # "CS" si on veut l'état civil apres 1796
    "pageNumber" : "1", # location qu'on veut trouver
}

# get
r = rq.post(
    places_url, 
    data=request, 
    cookies=dict(res.cookies),
)
pd.DataFrame(r.json()['response']['docs'])
```
