# Solr 

L'objectif de ce notebook est de parvenir à définir un outil "arch_search" qui puisse effectuer une recherche dans les corpus de agatha en utilisant pysolr.

```python
import pysolr
import pandas as pd
```

```python
############################################################################
## CA ICI, ca correspond aux résultats qui viennent de la recherche dans 
## 
############################################################################
host = "http://solr1.arch.local:8983/solr/ead" 
solr = pysolr.Solr(host)
res  = solr.search(q="gillard", df="ALL_TEXTS", wt="json")
df   = pd.DataFrame(res)
df.columns
```

```python
df.head()
```
